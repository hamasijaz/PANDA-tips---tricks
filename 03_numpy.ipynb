# Making an array from a data or another array

```python
a = np.arange(0, 222, 5)
a
```

```python
len(a)
```

```python
array1 = a[40:45]
array1
```

```python
a = np.array([1, 2, 3, 4, 5, 6])
b = np.array([7, 8, 9, 10, 11, 12])
c = np.array([13, 14, 15, 16, 17, 18])
d = np.hstack((a, b, c))
print(d)
e = np.vstack((a, b, c))
e
```

# Matrix or matrices

```python
# create a matrix 2 x 3
a = np.array([[1, 2, 3], [4, 5, 6]])
a
```

```python
# create a matrix in big dimension
a = np.arange(0, 200, 5).reshape(5, 8)
a
```

```python
# slicing of matrix
a[3:5, 2:4]
```

```python
a.max(), a.min(), sum(a)
```

```python
a
```

```python
a.max(axis=0), # only for column 
```

```python
a.min(axis=1) # only for row
```
