# 15-Plots in Plotly 

```python
# import library
import plotly.express as px
import pandas as pd
import numpy as np
import plotly.graph_objects as go
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.io as pio
pio.renderers.default = "vscode"

```

```python
df = sns.load_dataset('iris')
df.head()
```

```python
df.info()
```

```python
df.columns
```

# 01. Scatter plot

```python
# scatter plot
fig1 = px.scatter(df, x='sepal_length', y='sepal_width', color='species')
fig1.show()
# save the plot
fig1.write_image("./output/figu1.png", scale=3)
```

# 2. Line Plot

```python
fig2 = px.line(df, x='sepal_length', y='sepal_width', color='species', facet_col='species')
fig2.show()
# save the plot
fig2.write_image("./output/figu2.png", scale=3)
```

# 4. Bar plot

```python
from matplotlib import scale


fig4 = px.box(df, x='species', y='sepal_width', color='species')
fig4.show()
# save the plot 
fig4.write_image("./output/figu4.png", scale=3)
```

# 5. Violine plot

```python
fig5 = px.violin(df, x='species', y='sepal_width', color='species', box=True)
fig5.show()
# save the plot 
fig5.write_image("./output/figu5.png", scale=3)
```

# 6. Histogram 

```python
# histogram plot
fig6 = px.histogram(df, x='sepal_width', nbins=20, color='species', marginal="box")
fig6.show()
# save the plot 
fig6.write_image("./output/figu6.png", scale=3)
```

# 7. Pie chart

```python
# pie chart plot
fig7 = px.pie(df, values='sepal_width', names='species', hole=0.3)
fig7.show()
# save the plot 
fig7.write_image("./output/figu7.png", scale=3)
```

# 8. Scatter 3d plot


```python
# scatter 3d plot 
fig8 = px.scatter_3d(df, x='sepal_length', y='sepal_width', z='petal_length', color='species')
fig8.show()
# save the plot 
fig8.write_image("./output/figu8.png", scale=3)
```

# 9. Area chart 

```python
# area chart 
df_area = df.sort_values(by='sepal_length')
fig9 = px.area(df, x='sepal_length', y='sepal_width', color='species')
fig9.show()
# save the plot 
fig9.write_image("./output/figu9.png", scale=3)
```

# 10. Bubble chart

```python
fig10 = px.scatter(df, x='sepal_length', y='sepal_width', color='species', size='petal_width')
fig10.show()
# save the plot 
fig10.write_image("./output/figu10.png", scale=3)
```

# 11. sunburst chart 

```python
# sunburst chart plot
df_sun = df.groupby(['species', 'petal_length']).size().reset_index(name='counts')
fig11 = px.sunburst(df_sun, path=['species', 'petal_length'], values='counts')
fig11.show()
# save the plot 
fig11.write_image("./output/figu11.png", scale=3)
```

# 12. Parallel coordinates plot


```python
import pandas as pd
import plotly.express as px

# convert 'species' column to categorical type
df['species'] = pd.Categorical(df['species'])

# create a dictionary mapping species to colors
species_colors = {
    'setosa': 'red',
    'versicolor': 'blue',
    'virginica': 'green'
}

# map the species to colors
df['color'] = df['species'].map(species_colors)

# parallel coordinates plot
fig12 = px.parallel_coordinates(df, color='color')
fig12.show()
```

```python
import pandas as pd

# convert 'species' column to categorical type
df['species'] = pd.Categorical(df['species'])

# Check unique values in the 'species' column
unique_species = df['species'].unique()
print(unique_species)
```

```python
# add numeric speies_id based on species name 
df['species_id'] = df['species'].astype('category').cat.codes

# parallel coordinates plot
fig13 = px.parallel_coordinates(df, color='species_id', labels={'species_id': 'Species ID'}, color_continuous_scale='viridis')
fig13.show()
# save the plot 
fig13.write_image("./output/fig13.png", scale=3)
```

# 13. Density contour plot

```python
# Density contour plot
fig13 = px.density_contour(df, x='sepal_length', y='sepal_width', color='species')
fig13.show()
# save the plot 
fig13.write_image("./output/fig14.png", scale=3)
```

# 14. Ternery plot

```python
# Ternery plot
fig14 = px.scatter_ternary(df, a='sepal_length', b='sepal_width', c='petal_length', color='species')
fig14.show()
# save the plot 
fig14.write_image("./output/fig15.png", scale=3)
```

# 15. Polar chart (Radar chart)

```python
# Polar chart (Radar chart)
fig15 = px.line_polar(df, r='sepal_length', theta='species', line_close=True)
fig15.show()
# save the plot 
fig15.write_image("./output/fig16.png", scale=3)
```
